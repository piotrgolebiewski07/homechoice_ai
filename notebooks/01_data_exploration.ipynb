```python
def classify_offer(price_difference):
    if price_difference < -10:
        return "Atrakcyjna"
    elif price_difference <= 10:
        return "W średniej"
    else:
        return "Powyżej średniej"
```

```python
apartments_df["price_rating"] = apartments_df["price_difference_pct"].apply(classify_offer)
```
